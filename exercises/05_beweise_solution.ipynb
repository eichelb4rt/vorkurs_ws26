# Übung Beweise

## Aufgabe 1

Beweisen Sie direkt:

1. Das Produkt zweier ungeraden ganzen Zahlen ist ungerade.
2. Gegeben drei ganze Zahlen $a,b,c$: wenn $b$ durch $a$ teilbar ist, und $c$ durch $b$ teilbar, dann ist $c$ auch durch $a$ teilbar.

### Lösung

mit $k_i \in \mathbb{Z}$

1. Sei $a = 2\cdot k_1 + 1$ und $b = 2\cdot k_2 + 1$, dann gilt $a \cdot b = (2 k_1 + 1) (2 k_2 + 1) = 4 k_1 k_2 + 2k_1 + 2k_2 + 1 = 2 (2 k_1 k_2 + k_1 + k_2) + 1$.
2. Sei $b = a k_1$ und $c = b k_2$, dann ist $c = a k_1 k_2$.

## Aufgabe 2

Beweisen Sie durch Kontraposition:

1. Gegeben $x,y \in \mathbb{R}$ mit $x + y \geq 100$, dann muss $x \geq 50$ oder $y \geq 50$ gelten.
2. Gegeben $x,y \in \mathbb{R}$: wenn $x \cdot y$ irrational ist, dann muss $x$ oder $y$ irrational sein.

### Lösung

1. Die Kontraposition ist: Gegeben $x,y \in \mathbb{R}$ mit $x < 50$ und $y < 50$, dann muss $x + y < 100$ gelten. Die Kontraposition ist trivial, also gilt die ursprüngliche Aussage.
2. Die Kontraposition ist: Gegeben $x,y \in \mathbb{R}$: wenn $x$ und $y$ rational sind, ist $x \cdot y$ auch rational. Das können wir mit Multiplikationsregeln von Brüchen zeigen: sei $x = \frac{a_1}{b_1}$ und $y = \frac{a_2}{b_2}$, dann ist $x \cdot y = \frac{a_1}{b_1} \cdot \frac{a_2}{b_2} = \frac{a_1 \cdot a_2}{b_1 \cdot b_2}$ und damit rational.

## Aufgabe 3

Beweisen Sie mit Widerspruchsbeweis:

1. Es gibt keine größte natürliche Zahl.
2. Gegeben $x,y \in \mathbb{R}$: wenn $x$ rational ist und $y$ irrational, dann ist $x+y$ irrational.
3. Es gibt keine größte Primzahl.

### Lösung

1. Sei $n$ die größte natürliche Zahl. Dann ist $n + 1$ eine natürliche Zahl, die größer als $n$ ist. Das ist eine Verletzung der Annahme, dass $n$ die größte natürliche Zahl ist, und führt somit zum Widerspruch.
2. Sei $x + y$ rational, dann wäre $y = (x + y) - x$ eine Differenz zweier rationaler Zahlen und somit selbst rational. Das ist ein Widerspruch mit der Annahme, dass $y$ irrational ist.
3. Sei $p_1, \dots, p_n$ die Liste aller Primzahlen. Da $q = \prod_{i=1}^n p_i + 1$ nicht in der Liste ist, kann $q$ keine Primzahl sein. Somit muss sie einen Primfaktor in der Liste haben. Das ist im Widerspruch dazu, dass $q$ nach Konstruktion teilerfremd zu $p_1, \dots, p_n$ ist.

## Aufgabe 4

Finde Gegenbeispiele zu den folgenden Aussagen:

1. Alle Primzahlen sind ungerade.
2. Für alle reellen Zahlen gilt $x^2 > x$.
3. Gegeben $a,b,c \in \mathbb{N}$: wenn $b \cdot c$ durch $a$ teilbar ist, dann ist $b$ oder $c$ durch $a$ teilbar.

### Lösung

1. Die 2 ist eine gerade Primzahl
2. 0.5^2 < 0.5
3. $a = 6, b = 10, c = 15$. Somit ist $b \cdot c = 150 = 25 \cdot a$, aber weder $b$ noch $c$ durch $a$ teilbar.
