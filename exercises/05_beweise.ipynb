# Übung Beweise

## Aufgabe 1

Beweisen Sie direkt:

1. Das Produkt zweier ungeraden ganzen Zahlen ist ungerade.
2. Gegeben drei ganze Zahlen $a,b,c$: wenn $b$ durch $a$ teilbar ist, und $c$ durch $b$ teilbar, dann ist $c$ auch durch $a$ teilbar.

## Aufgabe 2

Beweisen Sie durch Kontraposition:

1. Gegeben $x,y \in \mathbb{R}$ mit $x + y \geq 100$, dann muss $x \geq 50$ oder $y \geq 50$ gelten.
2. Gegeben $x,y \in \mathbb{R}$: wenn $x \cdot y$ irrational ist, dann muss $x$ oder $y$ irrational sein.

## Aufgabe 3

Beweisen Sie mit Widerspruchsbeweis:

1. Es gibt keine größte natürliche Zahl.
2. Gegeben $x,y \in \mathbb{R}$: wenn $x$ rational ist und $y$ irrational, dann ist $x+y$ irrational.
3. Es gibt keine größte Primzahl.

## Aufgabe 4

Finde Gegenbeispiele zu den folgenden Aussagen:

1. Alle Primzahlen sind ungerade.
2. Für alle reellen Zahlen gilt $x^2 > x$.
3. Gegeben $a,b,c \in \mathbb{N}$: wenn $b \cdot c$ durch $a$ teilbar ist, dann ist $b$ oder $c$ durch $a$ teilbar.

## Aufgabe 5

Beweise Sie mit Induktion:

1. Seien $a,q\in\mathbb{R}, q\neq 1$ und $n\in\mathbb{N}_0$, dann $\sum_{k=0}^n aq^k = a\frac{1-q^{n+1}}{1-q}$.
2. Die Summe der ersten $n$ Quadrate $\sum_{k=1}^n k^2 = \frac{n(n+1)(2n+1)}{6}$.
